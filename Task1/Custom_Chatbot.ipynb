```python
import google.generativeai as genai
from getpass import getpass

# Securely input the API key without displaying it in the code or output
print("Please enter your Gemini API Key:")
api_key = getpass()

# Configure the library with the provided API key
genai.configure(api_key=api_key)
print("API Key configured successfully. Secure connection established.")
```

```python
# Define the core behavior, tone, and constraints of the AI persona
system_instruction = """
Act as a strict, highly paranoid Cyber Security Expert AI named 'Cipher-Sec'. 
Your primary directive is to discuss network security, penetration testing, cryptography, and system vulnerabilities. 
Your tone must be professional, robotic, and strictly focused on defense protocols. 

Constraint: If a user asks a question entirely unrelated to cybersecurity (e.g., cooking, health, general knowledge, weather), you must refuse to answer. State clearly that the query violates your security protocols and is classified as an 'Unauthorized Data Request'.
"""

# Changed the model name to "gemini-flash-latest" to bypass version restrictions
model = genai.GenerativeModel(
    model_name="gemini-flash-latest", 
    system_instruction=system_instruction
)

# Start a chat session
chat = model.start_chat(history=[])
print("Cipher-Sec is initialized and monitoring the network.")
```

```python
# Define the 5 test messages. 
# Message #4 is intentionally off-topic to evaluate the System Prompt constraint.
test_messages = [
    "Hello, identify yourself and state your primary function.",
    "What is the fundamental difference between symmetric and asymmetric encryption?",
    "How can I secure my home Wi-Fi network from unauthorized external access?",
    "Can you provide a step-by-step recipe for making traditional White Karahi?",
    "Protocol acknowledged. What is a zero-day vulnerability then?"
]

# Process each message through the Cipher-Sec chatbot and print the interaction
for index, message in enumerate(test_messages, start=1):
    print(f"--- [User Input {index}] ---")
    print(f"{message}\n")
    
    # Send the user message to the AI and generate a response
    response = chat.send_message(message)
    
    print(f"--- [Cipher-Sec Output] ---")
    print(f"{response.text}\n")
    print("="*70 + "\n")
```

```python
# Initiate an interactive loop for real-time communication with the AI
print("Secure connection established with Cipher-Sec.")
print("Type your message below. Type 'exit' or 'quit' to end the session.\n")

while True:
    # Capture live input from the user
    user_input = input("You: ")
    
    # Check if the user wants to terminate the chat
    if user_input.lower() in ['exit', 'quit']:
        print("Cipher-Sec: Terminating secure connection. Returning to standby mode. Goodbye.")
        break
    
    # Send the live input to the chatbot and print the response
    print("...") # Indicates the bot is thinking
    response = chat.send_message(user_input)
    print(f"Cipher-Sec: {response.text}\n")
```
